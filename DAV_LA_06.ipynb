```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

df = pd.read_excel("DAV_Project dataset.xlsx",
                   sheet_name="Ecommerce_1500")

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.strftime("%b")

print(df.head())
```

```python
fig = px.histogram(
    df,
    x="Revenue",
    title="Distribution of E-commerce Revenue",
    nbins=30
)

fig.update_xaxes(rangeslider_visible=True)
fig.show()
```

```python
fig = px.scatter(
    df,
    x="Quantity",
    y="Revenue",
    color="Product_Category",
    hover_data=["Order_ID", "Region"],
    title="Quantity vs Revenue"
)

fig.show()
```

```python
fig = px.box(
    df,
    x="Customer_Type",
    y="Revenue",
    color="Customer_Type",
    points="all",
    title="Revenue by Customer Type"
)

fig.show()
```

```python
fig = px.scatter_3d(
    df,
    x="Quantity",
    y="Unit_Price",
    z="Revenue",
    color="Product_Category",
    title="3D E-commerce Sales Analysis"
)

fig.show()
```

```python
df["Month_Number"] = df["Order_Date"].dt.month

df = df.sort_values("Order_Date")

fig = px.scatter(
    df,
    x="Quantity",
    y="Revenue",
    color="Product_Category",
    size="Unit_Price",
    animation_frame="Month_Number",
    hover_data=["Order_ID", "Order_Date"],
    range_x=[0, df["Quantity"].max() + 1],
    range_y=[0, df["Revenue"].max() * 1.1],
    title="Animated E-commerce Sales by Month"
)

fig.show()
```

```python
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=df["Order_ID"],
    y=df["Revenue"],
    mode="markers",
    name="Revenue"
))

fig.update_layout(
    title="E-commerce Metric Analysis",
    xaxis_title="Order ID",
    yaxis_title="Revenue",
    updatemenus=[
        dict(
            buttons=[
                dict(
                    label="Revenue",
                    method="update",
                    args=[
                        {"y": [df["Revenue"]]},
                        {"yaxis": {"title": "Revenue"}}
                    ]
                ),
                dict(
                    label="Quantity",
                    method="update",
                    args=[
                        {"y": [df["Quantity"]]},
                        {"yaxis": {"title": "Quantity"}}
                    ]
                ),
                dict(
                    label="Delivery Days",
                    method="update",
                    args=[
                        {"y": [df["Delivery_Days"]]},
                        {"yaxis": {"title": "Delivery Days"}}
                    ]
                )
            ],
            direction="down",
            x=1.0,
            y=1.15
        )
    ]
)

fig.show()
```
